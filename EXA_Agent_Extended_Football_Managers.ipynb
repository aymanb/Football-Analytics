```python
# ── CONFIG ── Edit this cell to change companies or the research task ──────────────

MANAGERS = [
    {"manager": "Jose Mourinho"},
    {"manager": "Pep Guardiola"},
]

RESEARCH_QUERY = """
For each input manager, produce a tactical and career research brief covering:
- Tactical philosophy and preferred system (formation, pressing style, build-up approach, defensive shape)
- Career history and key clubs managed, with notable results and trophy haul
- Signature strengths and what makes them distinctive as a manager
- Known weaknesses or recurring criticisms (e.g. player relationships, late-career patterns)
- How their style compares to peers and rivals — key contrasts and competitive rivalries
- Source URLs backing the research
"""
```

```python
from exa_py import Exa
from google.colab import userdata
import json
Exa_key = userdata.get('EXA_KEY')
exa = Exa(api_key=Exa_key)
```

```python
run = exa.beta.agent.runs.create(
    betas=["agent-2026-05-07"],
    query=RESEARCH_QUERY,
    input={"data": MANAGERS},
    output_schema={
        "type": "object",
        "required": ["reports"],
        "properties": {
            "reports": {
                "type": "array",
                "items": {
                    "type": "object",
                    "required": [
                        "manager",
                        "overview",
                        "tacticalSystem",
                        "careerHistory",
                        "keyStrengths",
                        "weaknesses",
                        "peerComparisons",
                        "sourceUrls",
                    ],
                    "properties": {
                        "manager": {"type": "string"},
                        "overview": {"type": "string"},
                        "tacticalSystem": {
                            "type": "array",
                            "items": {"type": "string"},
                            "minItems": 1,
                        },
                        "careerHistory": {
                            "type": "array",
                            "items": {"type": "string"},
                            "minItems": 1,
                        },
                        "keyStrengths": {
                            "type": "array",
                            "items": {"type": "string"},
                            "minItems": 1,
                        },
                        "weaknesses": {
                            "type": "array",
                            "items": {"type": "string"},
                            "minItems": 1,
                        },
                        "peerComparisons": {
                            "type": "array",
                            "items": {"type": "string"},
                            "minItems": 1,
                        },
                        "sourceUrls": {
                            "type": "array",
                            "items": {"type": "string", "format": "uri"},
                            "minItems": 1,
                        },
                    },
                },
            }
        },
    },
)

print(json.dumps(run.model_dump(), indent=2))
```

```python
completed_run = exa.beta.agent.runs.poll_until_finished(
    run.id, betas=["agent-2026-05-07"]
)
print(completed_run.output.structured if completed_run.output else None)
```

```python
from IPython.display import display, Markdown

if completed_run.output and completed_run.output.structured:
    reports = completed_run.output.structured.get('reports', [])
    for report in reports:
        display(Markdown(f"## {report['manager']}"))
        display(Markdown(f"**Overview:** {report['overview']}"))

        display(Markdown("**Tactical System:**"))
        display(Markdown("\n".join(f"- {t}" for t in report['tacticalSystem'])))

        display(Markdown("**Career History:**"))
        display(Markdown("\n".join(f"- {h}" for h in report['careerHistory'])))

        display(Markdown("**Key Strengths:**"))
        display(Markdown("\n".join(f"- {s}" for s in report['keyStrengths'])))

        display(Markdown("**Weaknesses:**"))
        display(Markdown("\n".join(f"- {w}" for w in report['weaknesses'])))

        display(Markdown("**Peer Comparisons:**"))
        display(Markdown("\n".join(f"- {p}" for p in report['peerComparisons'])))

        display(Markdown("**Sources:**"))
        display(Markdown("\n".join(f"- [{url}]({url})" for url in report['sourceUrls'])))

        display(Markdown("---"))
else:
    print("No structured output found to display.")
```
